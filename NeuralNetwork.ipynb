```python
import Preprocessing
import tensorflow as tf
import tensorflow.keras as keras
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, accuracy_score, recall_score, precision_score
```

```python
nnmodel = tf.keras.Sequential([
    tf.keras.layers.Dense(16, activation = 'relu', input_shape = [7,]),
    tf.keras.layers.Dense(8, activation = 'relu'),
    tf.keras.layers.Dense(1, activation = 'sigmoid')
])
opt = tf.keras.optimizers.Adam(learning_rate = 1e-3)
nnmodel.compile(loss = 'binary_crossentropy', optimizer = opt, metrics = ['acc'])
```

```python
x = Preprocessing.x_scaled
y = Preprocessing.y
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=123)
```

```python
nnmodel.fit(x_train,y_train,epochs=100,batch_size=32)
y_fitted = nnmodel.predict(x_test)
y_fitted = [1 if i>0.5 else 0 for i in y_fitted]
```

```python
cm = confusion_matrix(y_test,y_fitted)
print("Confusion matrix: ")
print(cm)
accuracy = accuracy_score(y_test, y_fitted)
print("Accuracy Score: ", accuracy)
recall = recall_score(y_test, y_fitted)
print("Recall Score: ", recall)
precision = precision_score(y_test, y_fitted)
print("Precision Score: ", precision)
```
