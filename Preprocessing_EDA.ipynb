```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
titanic = pd.read_csv('Titanic.csv')
df = pd.DataFrame(titanic)
excluded = ['PassengerId', 'Name', 'Ticket', 'Cabin']
df = df.drop(columns=excluded)
df = df.dropna(subset=['Age'])
df = df.dropna(subset=['Embarked'])
remained = df.shape[0]
print('The number of remained observations is', remained)
print(df.head(5))
```

```python
df['Survived'] = df['Survived'].astype(str)
df['Pclass'] = df['Pclass'].astype(str)
df['Sex'] = df['Sex'].astype(str)
df['Embarked'] = df['Embarked'].astype(str)
```

```python
sns.countplot(x=df['Survived'], color = 'steelblue')
plt.title("Survival Distribution")
plt.show()
survival_counts = df['Survived'].value_counts()
print(f"Number of passengers who did not survive: {survival_counts.iloc[0]}")
print(f"Number of passengers who survived: {survival_counts.iloc[1]}")
```

```python
sns.countplot(df,x=df['Pclass'], hue=df['Survived'], order=['1','2','3'])
plt.title("Survival Distribution regarding to Class")
plt.xlabel("Passenger Class")
plt.ylabel("Count")
plt.show()

pclass_counts = df['Pclass'].value_counts()
print(f"Number of passengers in the first class: {pclass_counts.iloc[2]}")
print(f"Number of passengers in the second class: {pclass_counts.iloc[1]}")
print(f"Number of passengers in the third class: {pclass_counts.iloc[0]}")
```

```python
sns.countplot(df, x = df['Sex'], hue = df['Survived'])
plt.title("Survival Distribution regarding to Sex")
plt.xlabel("Sex")
plt.ylabel("Count")

sex_counts = df['Sex'].value_counts()
print(f"Number of male passengers: {sex_counts.iloc[0]}")
print(f"Number of female passengers: {sex_counts.iloc[1]}")
```

```python
bins = list(range(0,100,10))
labels = ['0-10', '10-19', '20-29', '30-39', '40-49', '50-59', '60-69', '70-79', '80-89']
df['AgeGroup'] = pd.cut(df['Age'], bins = bins, labels = labels, right = False)
df['AgeGroup'] = df['AgeGroup'].astype(str)

sns.countplot(df, x = 'AgeGroup', hue = 'Survived')
plt.title("Survival Distribution Regarding Age Group")
plt.xlabel("Age Group")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()

age_group_counts = df['AgeGroup'].value_counts().sort_index()
print(f"Number passengers under 10 years old: {age_group_counts.iloc[0]}")
print(f"Number of passengers between 10 and 19: {age_group_counts.iloc[1]}")
print(f"Number of passengers between 20 and 29: {age_group_counts.iloc[2]}")
print(f"Number of passengers between 30 and 39: {age_group_counts.iloc[3]}")
print(f"Number of passengers between 40 and 49: {age_group_counts.iloc[4]}")
print(f"Number of passengers between 50 and 59: {age_group_counts.iloc[5]}")
print(f"Number of passengers between 60 and 69: {age_group_counts.iloc[6]}")
print(f"Number of passengers between 70 and 79: {age_group_counts.iloc[7]}")
print(f"Number of passengers between 80 and 89: {age_group_counts.iloc[8]}")
```

```python
df['SibSp'] = df['SibSp'].astype(str)
sns.countplot(df, x = 'SibSp', hue = 'Survived', order=['0','1','2','3','4','5'])
plt.title("Survival Distribution Regarding the Number of Siblings or Spouses Onboard")
plt.xlabel("Number of Siblings and Spouses Onboard")
plt.ylabel("Count")
plt.show()

SibSp_counts = df['SibSp'].value_counts().sort_index()

print(f"Number of passengers without siblings and spouses onboard: {SibSp_counts.iloc[0]}")
print(f"Number of passengers with one sibling or spouse onboard: {SibSp_counts.iloc[1]}")
print(f"Number of passengers with two siblings or spouses onboard: {SibSp_counts.iloc[2]}")
print(f"Number of passengers with three siblings or spouses onboard: {SibSp_counts.iloc[3]}")
print(f"Number of passengers with four siblings or spouses onboard: {SibSp_counts.iloc[4]}")
print(f"Number of passengers with five siblings or spouses onboard: {SibSp_counts.iloc[5]}")
```

```python
df['Parch'] = df['Parch'].astype(str)
sns.countplot(df, x = 'Parch', hue = 'Survived', order=['0','1','2','3','4','5','6'])
plt.title("Survival Distribution Regarding the Number of Parents or Children Onboard")
plt.xlabel("Number of Parents or Children Onboard")
plt.ylabel("Count")
plt.show()

Parch_counts = df['Parch'].value_counts().sort_index()

print(f"Number of passengers without parents and children onboard: {Parch_counts.iloc[0]}")
print(f"Number of passengers with one parent or child onboard: {Parch_counts.iloc[1]}")
print(f"Number of passengers with two parents or children onboard: {Parch_counts.iloc[2]}")
print(f"Number of passengers with three parents or children onboard: {Parch_counts.iloc[3]}")
print(f"Number of passengers with four parents or children onboard: {Parch_counts.iloc[4]}")
print(f"Number of passengers with five parents or children onboard: {Parch_counts.iloc[5]}")
print(f"Number of passengers with five parents or children onboard: {Parch_counts.iloc[6]}")
```

```python
df['FareGroup'] = np.where(df['Fare'] < 50, '<50', '>50')
sns.countplot(df, x = 'FareGroup', hue = 'Survived')
plt.title("Survival Distribution Regarding Fare")
plt.xlabel("Fare Group")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()

fare_group_counts = df['FareGroup'].value_counts().sort_index()
print(f"Passenger fare less than 50: {fare_group_counts.iloc[0]}")
print(f"Passenger fare more than 50: {fare_group_counts.iloc[1]}")
```
