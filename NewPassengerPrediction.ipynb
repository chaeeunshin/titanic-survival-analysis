```python
import numpy as np
import time
import warnings
import Preprocessing
from logistic_model import LogisticModel
from rf_model import RandomForestModel
from nn_model import NeuralNetworkModel
```

```python
isClassValid = False
isSexValid = False
isAgeValid = False
isSibSpValid = False
isParchValid = False
isFareValid = False
isEmbarkedValid = False
isModelValid = False

print("This program is designed to predict a new passenger's survival in Titanic.")
time.sleep(1)
new_passenger_Class = input("Please enter the new passenger's class. 1 for the first class, 2 for the second, and 3 for third:")
if new_passenger_Class.isdigit() and int(new_passenger_Class) >= 1 and int(new_passenger_Class) <= 3:
    isClassValid = True
else: 
    isClassValid = False
while not isClassValid:
    new_passenger_Class = input("Invalid input. Please write in an integer. 1 for the first class, 2 for the second, and 3 for third:")
    if new_passenger_Class.isdigit() and int(new_passenger_Class) >= 1 and int(new_passenger_Class) <= 3:
        isClassValid = True
        break

time.sleep(1)
new_passenger_Sex = input("Please enter the new passenger's sex. 0 for male, 1 for female:")
if new_passenger_Sex.isdigit() and int(new_passenger_Sex) == 1 or int(new_passenger_Sex) == 0:
    isSexValid = True
else: 
    isSexValid = False
while not isSexValid:
    new_passenger_Sex = input("Invalid input. Please write in an integer. 1 for the first class, 2 for the second, and 3 for third:")
    if new_passenger_Sex.isdigit() and int(new_passenger_Sex) == 1 or int(new_passenger_Sex) == 0:
        isSexValid = True
        break

time.sleep(1)
new_passenger_Age = input("Please enter the new passenger's age.:")
if new_passenger_Age.isdigit() and int(new_passenger_Age)>=0:
    isAgeValid = True
else:
    isAgeValid = False
while not isAgeValid:
    new_passenger_Age = input("Invalid input. Please write in a positive integer.")
    if new_passenger_Age.isdigit() and int(new_passenger_Age) >= 0:
        isAgeValid = True
        break

time.sleep(1)
new_passenger_SibSp = input("Enter the number of siblings/spouses aboard: ")
if new_passenger_SibSp.isdigit() and int(new_passenger_SibSp) >= 0:
    isSibSpValid = True
else:
    isSibSpValid = False
while not isSibSpValid:
    new_passenger_SibSp = input("Invalid input. Please write in a positive integer.")
    if new_passenger_SibSp.isdigit() and int(new_passenger_SibSp) >= 0:
        isSibSpValid = True
        break

time.sleep(1)
new_passenger_Parch = input("Enter the number of parents/children aboard: ")
if new_passenger_Parch.isdigit() and int(new_passenger_Parch) >= 0:
    isParchValid = True
else:
    isParchValid = False
while not isParchValid:
    new_passenger_Parch = input("Invalid input. Please write in a positive integer.")
    if new_passenger_Parch.isdigit() and int(new_passenger_Parch) >= 0:
        isParchValid = True
        break

time.sleep(1)
new_passenger_Fare = input("Enter the passenger fare, in British Pound: ")
if float(new_passenger_Fare) >= 0:
    isFareValid = True
else:
    isFareValid = False
while not isFareValid:
    new_passenger_Fare = input("Invalid input. Please write in a positive numeric value.")
    if float(new_passenger_Fare) >= 0:
        isFareValid = True
        break
        
time.sleep(1)
new_passenger_Embarked = input("Enter port of embarkation. C for Cherbourg, Q for Queenstown, S for Southampton.: ")
if new_passenger_Embarked == 'C' or new_passenger_Embarked == 'Q' or new_passenger_Embarked == 'S':
    isEmbarkedValid = True
    if new_passenger_Embarked == 'C':
        new_passenger_Embarked = 0
    elif new_passenger_Embarked == 'Q':
        new_passenger_Embarked = 1
    elif new_passenger_Embarked == 'S':
        new_passenger_Embarked = 2
else:
    isEmbarkedValid = False
while not isEmbarkedValid:
    new_passenger_Embarked = input("Invalid input. Please enter either C, Q, or S in uppercase.: ")
    if new_passenger_Embarked == 'C' or new_passenger_Embarked == 'Q' or new_passenger_Embarked == 'S':
        isEmbarkedValid = True
        break

warnings.filterwarnings("ignore")
new_passenger_Info = np.array([int(new_passenger_Class), int(new_passenger_Sex),
                               int(new_passenger_Age), int(new_passenger_SibSp),
                               int(new_passenger_Parch), float(new_passenger_Fare), 
                               int(new_passenger_Embarked)])
new_passenger_Info = new_passenger_Info.reshape(1,-1)

predictionModel = input("Select a model for prediction. 0 for general regression, 1 for random forest, 2 for neural network.:")
if predictionModel.isdigit() and int(predictionModel) == 0 or int(predictionModel) == 1 or int(predictionModel) == 2:
    isModelValid =True
    if int(predictionModel) == 0:
        new_passenger_Survived = LogisticModel.predict(new_passenger_Info)
    elif int(predictionModel) == 1:
        new_passenger_Survived = RandomForestModel.predict(new_passenger_Info)
    elif int(predictionModel) == 2:
        new_passenger_Survived = NeuralNetworkModel.predict(new_passenger_Info)
    
else:
    while not isModelValid:
        predictionModel = input("Invalid input for model selection. Please enter either 0, 1, or 2:")
        if predictionModel.isdigit() and int(predictionModel) == 0 or int(predictionModel) == 1 or int(predictionModel) == 2:
            isModelValid = True
            break
    
if new_passenger_Survived >= 0.5:
    print("The new passenger is likely to survive.")
else:
    print("The new passenger is not likely to survive.")
```
